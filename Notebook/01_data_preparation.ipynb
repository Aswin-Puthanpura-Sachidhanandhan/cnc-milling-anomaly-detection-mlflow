```python
DATASET_ID = "manufuturetoday/multi-sensor-for-metal-milling-anomaly"

print(DATASET_ID)
```

```python
import subprocess

result = subprocess.run(
    [
        "kaggle",
        "datasets",
        "files",
        "-d",
        DATASET_ID,
        "--page-size",
        "10000"
    ],
    capture_output=True,
    text=True
)

print("Return code:", result.returncode)
print(result.stdout[:10000])

if result.stderr:
    print("Error:")
    print(result.stderr)
```

```python
file_listing = result.stdout

csv_files = []

for line in file_listing.splitlines():
    line = line.strip()

    if not line or line.startswith("name") or line.startswith("---"):
        continue

    filename = line.split()[0]

    if filename.lower().endswith(".csv"):
        csv_files.append(filename)

print("Number of CSV files found:", len(csv_files))

for filename in csv_files[:30]:
    print(filename)
```

```python
import os
import subprocess

DOWNLOAD_FOLDER = "/tmp/msm_csv"

os.makedirs(DOWNLOAD_FOLDER, exist_ok=True)

success = 0
failed = 0

for file in csv_files:
    print(f"Downloading: {file}")

    result = subprocess.run(
        [
            "kaggle",
            "datasets",
            "download",
            "-d",
            DATASET_ID,
            "-f",
            file,
            "-p",
            DOWNLOAD_FOLDER,
            "--unzip"
        ],
        capture_output=True,
        text=True
    )

    if result.returncode == 0:
        success += 1
    else:
        failed += 1
        print(result.stderr)

print("\nFinished!")
print(f"Downloaded: {success}")
print(f"Failed: {failed}")
```

```python
import os

csv_count = 0
total_size = 0

for root, dirs, files in os.walk(DOWNLOAD_FOLDER):
    for f in files:
        csv_count += 1
        total_size += os.path.getsize(os.path.join(root, f))

print("CSV Files:", csv_count)
print(f"Total Size: {total_size/(1024*1024):.2f} MB")
```

```python
import os
import pandas as pd

mtc_remote_files = [
    path for path in csv_files
    if path.lower().endswith("_mtc.csv")
]

pairing_rows = []

for mtc_remote in mtc_remote_files:
    parent_folder = os.path.dirname(mtc_remote)

    local_mtc = os.path.join(
        "/tmp/msm_csv",
        os.path.basename(mtc_remote)
    )

    local_label = os.path.join(
        "/tmp/msm_labels",
        parent_folder.replace("/", "_"),
        "label.csv"
    )

    pairing_rows.append({
        "dataset_folder": parent_folder,
        "mtc_remote": mtc_remote,
        "local_mtc": local_mtc,
        "local_label": local_label,
        "mtc_exists": os.path.exists(local_mtc),
        "label_exists": os.path.exists(local_label)
    })

pairing_df = pd.DataFrame(pairing_rows)

display(pairing_df)
```

```python
print("Total MTC entries:", len(pairing_df))
print("Local MTC files found:", pairing_df["mtc_exists"].sum())
print("Local label files found:", pairing_df["label_exists"].sum())

pairing_df["complete_pair"] = (
    pairing_df["mtc_exists"]
    & pairing_df["label_exists"]
)

print("Complete pairs:", pairing_df["complete_pair"].sum())
```

```python
alignment_results = []

for _, row in pairing_df[pairing_df["complete_pair"]].iterrows():

    try:
        mtc_sample = pd.read_csv(
            row["local_mtc"],
            usecols=["time"]
        )

        label_sample = pd.read_csv(
            row["local_label"]
        )

        label_sample.columns = label_sample.columns.str.strip()

        aligned = (
            label_sample["start"].min() >= mtc_sample["time"].min()
            and label_sample["end"].max() <= mtc_sample["time"].max()
        )

        alignment_results.append({
            "dataset_folder": row["dataset_folder"],
            "mtc_file": os.path.basename(row["local_mtc"]),
            "mtc_max_time": mtc_sample["time"].max(),
            "label_max_time": label_sample["end"].max(),
            "aligned": aligned
        })

    except Exception as error:
        alignment_results.append({
            "dataset_folder": row["dataset_folder"],
            "mtc_file": os.path.basename(row["local_mtc"]),
            "mtc_max_time": None,
            "label_max_time": None,
            "aligned": False,
            "error": str(error)
        })

alignment_df = pd.DataFrame(alignment_results)

display(alignment_df)
```

```python
import os
import numpy as np
import pandas as pd

all_refined_rows = []

for _, pair in pairing_df[pairing_df["complete_pair"]].iterrows():

    mtc_path = pair["local_mtc"]
    label_path = pair["local_label"]
    dataset_folder = pair["dataset_folder"]

    mtc = pd.read_csv(mtc_path)
    label = pd.read_csv(label_path)

    mtc.columns = mtc.columns.str.strip()
    label.columns = label.columns.str.strip()

    # Binary classification target:
    # 1 = anomaly
    # 0 = normal
    label["target"] = (label["label"] == 1).astype(int)

    numeric_mtc_columns = [
        column
        for column in mtc.select_dtypes(include=np.number).columns
        if column != "time"
        and mtc[column].nunique(dropna=True) > 1
    ]

    metadata_columns = [
        column
        for column in ["N", "ap", "ae", "F", "Z", "D"]
        if column in label.columns
    ]

    for segment_id, segment_info in label.iterrows():

        segment = mtc.loc[
            (mtc["time"] >= segment_info["start"])
            & (mtc["time"] <= segment_info["end"])
        ]

        if len(segment) < 2:
            continue

        refined_row = {
            "dataset_folder": dataset_folder,
            "mtc_file": os.path.basename(mtc_path),
            "segment_id": segment_id,
            "start": segment_info["start"],
            "end": segment_info["end"],
            "duration": segment_info["end"] - segment_info["start"],
            "raw_label": int(segment_info["label"]),
            "target": int(segment_info["target"]),
            "mtc_sample_count": len(segment),
        }

        if "direction" in label.columns:
            refined_row["direction_up"] = int(
                str(segment_info["direction"]).strip().upper() == "UP"
            )

        for column in metadata_columns:
            refined_row[column] = segment_info[column]

        for column in numeric_mtc_columns:

            values = (
                pd.to_numeric(segment[column], errors="coerce")
                .replace([np.inf, -np.inf], np.nan)
                .dropna()
            )

            if values.empty:
                continue

            refined_row[f"{column}_mean"] = values.mean()
            refined_row[f"{column}_std"] = values.std(ddof=0)
            refined_row[f"{column}_min"] = values.min()
            refined_row[f"{column}_max"] = values.max()
            refined_row[f"{column}_range"] = values.max() - values.min()

        all_refined_rows.append(refined_row)

refined_df = pd.DataFrame(all_refined_rows)

print("Refined dataset shape:", refined_df.shape)
display(refined_df.head())
```

```python
print("Rows:", len(refined_df))
print("Columns:", len(refined_df.columns))

print("\nTarget distribution:")
print(refined_df["target"].value_counts())

print("\nTarget percentage:")
print(
    refined_df["target"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nDuplicate rows:")
print(refined_df.duplicated().sum())

missing_percent = (
    refined_df.isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("\nTop columns with missing values:")
print(missing_percent[missing_percent > 0].head(20))
```

```python
identifier_columns = [
    "dataset_folder",
    "mtc_file",
    "segment_id",
    "start",
    "end",
    "raw_label",
    "target"
]

feature_columns = [
    column
    for column in refined_df.columns
    if column not in identifier_columns
]

all_missing_columns = [
    column
    for column in feature_columns
    if refined_df[column].isna().all()
]

constant_columns = [
    column
    for column in feature_columns
    if refined_df[column].nunique(dropna=True) <= 1
]

columns_to_remove = sorted(
    set(all_missing_columns + constant_columns)
)

print("All-missing columns:", len(all_missing_columns))
print("Constant columns:", len(constant_columns))
print("Total columns removed:", len(columns_to_remove))

refined_clean = refined_df.drop(columns=columns_to_remove)

print("Clean dataset shape:", refined_clean.shape)
```

```python
high_missing_columns = [
    column
    for column in refined_clean.columns
    if refined_clean[column].isna().mean() > 0.40
    and column not in identifier_columns
]

print("Columns above 40% missing:", len(high_missing_columns))

for column in high_missing_columns:
    print(
        column,
        round(refined_clean[column].isna().mean() * 100, 2),
        "%"
    )

refined_clean = refined_clean.drop(columns=high_missing_columns)

print("Shape after missing-value filtering:", refined_clean.shape)
```

```python
remaining_missing = (
    refined_clean.isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("Highest remaining missing percentages:")
print(remaining_missing[remaining_missing > 0].head(20))
```

```python
REFINED_PATH = "/tmp/refined_mtc_dataset.csv"

refined_clean.to_csv(REFINED_PATH, index=False)

print("Saved:", REFINED_PATH)
print("Rows:", refined_clean.shape[0])
print("Columns:", refined_clean.shape[1])
```
