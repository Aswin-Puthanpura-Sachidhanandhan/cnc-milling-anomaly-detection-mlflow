```python
import mlflow
import mlflow.sklearn
from mlflow import MlflowClient

REGISTERED_MODEL_NAME = "cnc_milling_anomaly_model"

client = MlflowClient()

champion = client.get_model_version_by_alias(
    name=REGISTERED_MODEL_NAME,
    alias="Champion"
)

print("Champion version:", champion.version)
print("Champion run ID:", champion.run_id)

champion_uri = (
    f"models:/{REGISTERED_MODEL_NAME}@Champion"
)

champion_model = mlflow.sklearn.load_model(
    champion_uri
)

print("Champion model loaded successfully.")
```

```python
import pandas as pd
import numpy as np

df = spark.table(
    "default.refined_mtc_dataset"
).toPandas()

identifier_columns = [
    "dataset_folder",
    "mtc_file",
    "segment_id",
    "start",
    "end",
    "raw_label"
]

X = df.drop(
    columns=identifier_columns + ["target"]
)

X = X.select_dtypes(include=np.number)

sample_input = X.head(1).copy()

print("Input shape:", sample_input.shape)
display(sample_input)
```

```python
prediction = champion_model.predict(
    sample_input
)[0]

probability = champion_model.predict_proba(
    sample_input
)[0, 1]

print("Prediction:", int(prediction))
print(
    "Result:",
    "Anomaly" if prediction == 1 else "Normal"
)
print("Anomaly probability:", float(probability))
```

```python
normal_index = df[df["target"] == 0].index[0]
anomaly_index = df[df["target"] == 1].index[0]

normal_input = X.loc[[normal_index]]
anomaly_input = X.loc[[anomaly_index]]

normal_prediction = champion_model.predict(normal_input)[0]
normal_probability = champion_model.predict_proba(normal_input)[0, 1]

anomaly_prediction = champion_model.predict(anomaly_input)[0]
anomaly_probability = champion_model.predict_proba(anomaly_input)[0, 1]

print("NORMAL EXAMPLE")
print("Actual:", 0)
print("Prediction:", int(normal_prediction))
print("Anomaly probability:", float(normal_probability))

print("\nANOMALY EXAMPLE")
print("Actual:", 1)
print("Prediction:", int(anomaly_prediction))
print("Anomaly probability:", float(anomaly_probability))
```

```python
import mlflow.deployments

ENDPOINT_NAME = "cnc-milling-anomaly-endpoint"

deployment_client = mlflow.deployments.get_deploy_client(
    "databricks"
)

request_payload = {
    "dataframe_split": {
        "columns": sample_input.columns.tolist(),
        "data": sample_input.astype(float).values.tolist()
    }
}

response = deployment_client.predict(
    endpoint=ENDPOINT_NAME,
    inputs=request_payload
)

print("Endpoint response:")
print(response)
```

```python
print("Endpoint:", ENDPOINT_NAME)
print("Actual label: 0")
print("Endpoint response:", response)
print("Interpretation: Normal")
```

```python
import pandas as pd

# Ensure the exact feature order expected by the model
anomaly_input = anomaly_input[sample_input.columns].copy()

# Convert NaN to Python None, which becomes JSON null
json_safe_anomaly_input = (
    anomaly_input
    .astype(object)
    .where(pd.notna(anomaly_input), None)
)

anomaly_payload = {
    "dataframe_split": {
        "columns": json_safe_anomaly_input.columns.tolist(),
        "data": json_safe_anomaly_input.values.tolist()
    }
}

print("Payload prepared successfully.")
```
